Handling os missing value 
using Titanic dataset

```python
import pandas as pd

df = pd.read_csv(r"C:\users\Nisarga\Desktop/DAV/Titanic-Dataset.csv")
print(df)
```

```python
# printing first five rows
print(df.head())
```

```python
print(df.isnull())   # identifies missing values,
print(df.isnull().sum())  # tells us how many missing values are present in each column. 
```

```python
# Mean Imputation
df['Age'] = df['Age'].fillna(df['Age'].mean())
print(df)
print(df['Age'].isnull().sum())
```

```python
# calculating mean separately

print("Mean Age:", df['Age'].mean())
print("Mean Fare:",df['Fare'].mean())
```

```python
# Median Imputation

print("Median Age:", df['Age'].median())
print("Median Fare:", df['Fare'].median())
```

```python
# before forward fill

print("Before Forward Fill:")
print(df[['PassengerId', 'Embarked']].head(10))
```

```python
# after forword fill

df['Embarked'] = df['Embarked'].ffill()
print("After Forward Fill:")
print(df[['PassengerId', 'Embarked']].head(999))
```

```python
# Label encoding

from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df['Sex_Encoded'] = encoder.fit_transform(df['Sex'])
print("After Label Encoding:")
print(df[['Sex', 'Sex_Encoded']].head(10))
print("\nClasses:")
print(encoder.classes_)
```

```python
# One-Hot Encoding

encoded_df = pd.get_dummies(
    df,
    columns=['Embarked'],
    dtype=int
)
print("After One-Hot Encoding:")
print(encoded_df[['Embarked_C', 'Embarked_Q', 'Embarked_S']].head(10))
```
